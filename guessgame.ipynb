```python
import random
```

```python
number=random.randint(1,50)

print("enter your choice:")
print("you have 3 chance")

for i in range(3):
  guess=int(input("enter you no:"))
  if guess==number:
    print("YOU WON")
    break

  elif guess>number:
    print("too high")

  else:
    print("too low")
else:
  print("YOU LOSEE..")
print("actual number:", number)
```
